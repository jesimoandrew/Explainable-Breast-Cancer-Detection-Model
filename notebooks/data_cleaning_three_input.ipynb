```python
import pandas as pd

raw_train_calc_df = pd.read_csv("../data_description/calc_case_description_train_set.csv")
raw_train_mass_df = pd.read_csv("../data_description/mass_case_description_train_set.csv")

raw_test_calc_df = pd.read_csv("../data_description/calc_case_description_test_set.csv")
raw_test_mass_df = pd.read_csv("../data_description/mass_case_description_test_set.csv")

raw_df = pd.concat([raw_train_calc_df, raw_train_mass_df, raw_test_calc_df, raw_test_mass_df], ignore_index=True)

raw_df.head()
```

```python
raw_df = raw_df[['patient_id', 'pathology', 'image file path', 'cropped image file path', 'ROI mask file path']]
```

```python
raw_df.head()
```

```python
raw_df['pathology'].value_counts()
```

```python
raw_df['pathology'] = raw_df['pathology'].map({'BENIGN': 0, 'MALIGNANT': 1, 'BENIGN_WITHOUT_CALLBACK': 0})

raw_df.head()
```

```python
target_columns = ['image file path', 'cropped image file path', 'ROI mask file path']


unique_vals = {col: raw_df[col].drop_duplicates().reset_index(drop=True) for col in target_columns}
for col, s in unique_vals.items():
    print(col, s.duplicated().sum())  # will all be 0

for col in target_columns:
    deduped = raw_df[col].drop_duplicates()
    print(col, "len:", len(deduped), "unique:", deduped.nunique(), "duplicated after dedup:", deduped.duplicated().sum())
```

```python
from sklearn.model_selection import StratifiedGroupKFold

# roughly 80/10/10 → use ~10 folds, take 1 fold as test
sgkf = StratifiedGroupKFold(n_splits=10, shuffle=True, random_state=42)

trainval_idx, test_idx = next(sgkf.split(
    raw_df,
    y=raw_df['pathology'],       # your label column
    groups=raw_df['patient_id']
))

trainval_df = raw_df.iloc[trainval_idx]
test_df     = raw_df.iloc[test_idx]

# repeat on trainval to carve out val (~10% of original ≈ 1/9 of trainval)
sgkf2 = StratifiedGroupKFold(n_splits=9, shuffle=True, random_state=42)

train_idx, val_idx = next(sgkf2.split(
    trainval_df,
    y=trainval_df['pathology'],
    groups=trainval_df['patient_id']
))

train_df = trainval_df.iloc[train_idx]
val_df   = trainval_df.iloc[val_idx]
```

```python
for name, d in [('train', train_df), ('val', val_df), ('test', test_df)]:
    print(name, len(d))
    print(d['pathology'].value_counts(normalize=True))
    print()

print("original")
print(raw_df['pathology'].value_counts(normalize=True))
```

```python
train_df.to_csv("../dataframes/three_input_train_df.csv", index=False)
test_df.to_csv("../dataframes/three_input_test_df.csv", index=False)
val_df.to_csv("../dataframes/three_input_val_df.csv", index=False)
```
